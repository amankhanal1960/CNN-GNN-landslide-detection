```python
import os
import h5py
import numpy as np

# Make sure 'root' is defined earlier in your notebook/script
mask_dir = (root / "output" / "masks")

mask_files = sorted([f for f in os.listdir(mask_dir) if f.endswith(".h5")])

# Dictionary to store {filename: positive_pixel_count}
pos_counts = {}
empty_files = []

for f in mask_files:
    path = os.path.join(mask_dir, f)
    with h5py.File(path, "r") as hf:
        m = hf["mask"][:]
        
        # Calculate the exact number of pixels > 0
        pos_pixels = np.sum(m > 0)
        
        if pos_pixels > 0:
            pos_counts[f] = pos_pixels
        else:
            empty_files.append(f)

print(f"Total Masks Checked: {len(mask_files)}")
print(f"Masks WITH Landslides (pos > 0): {len(pos_counts)}")
print(f"Empty / Blank Masks (pos == 0):  {len(empty_files)}")

# Print detailed counts and statistics if we found positive masks
if pos_counts:
    print("\n--- Pixel Count Statistics ---")
    counts = list(pos_counts.values())
    print(f"Highest pixel count in a tile: {np.max(counts)}")
    print(f"Lowest pixel count in a tile:  {np.min(counts)}")
    print(f"Average positive pixels/tile:  {np.mean(counts):.1f}")
    
    print("\n--- Detailed Tile Counts (First 20) ---")
    # Show the first 20 positive tiles to get a quick visual check
    for idx, (f, count) in enumerate(pos_counts.items()):
        if idx >= 20:
            break
        print(f"{f}: {count} positive pixels")
```
